```python
import keyword
word = input("Enter a word: ")
if keyword.iskeyword(word):
    print(f"'{word}' is a Python keyword.")
else:
    print(f"'{word}' is not a Python keyword.")
```

```python
def modify_list(lst):
    lst.append(4)

def modify_int(x):
    x += 1
    return x

my_list = [1, 2, 3]
my_int = 5

modify_list(my_list)
print("List after function call:", my_list)

new_int = modify_int(my_int)
print("Integer after function call:", new_int)
print("Original integer remains:", my_int)
```

```python
s = input("Enter something: ")

try:
    i = int(s)
    f = float(s)
    b = bool(s)

    print("Integer:", i)
    print("Float:", f)
    print("Boolean:", b)
except ValueError:
    print("Conversion failed. Input is not a number.")
```

```python
s = input("Enter a string: ")
reversed_str = ""
i = len(s) - 1

while i >= 0:
    reversed_str += s[i]
    i -= 1

print("Reversed string:", reversed_str)
```

```python
num = int(input("Enter a number: "))
factorial = 1
i = 1

while i <= num:
    factorial *= i
    i += 1

print("Factorial of", num, "is", factorial)
```
