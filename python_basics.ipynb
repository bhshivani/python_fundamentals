5. Mutability in Python means whether or not a value (like a list or string) can be changed after it is created.
•	If something is mutable, you can change it.
  Example: Lists are mutable.
Example in Python Jupyternotebook:
my_list = [1, 2, 3]
my_list[0] = 100  # List is changed to [100, 2, 3]

6. Lists are mutable because you can change, add, or remove items after the list is created.
Tuples are immutable because you cannot change them once they are created.
•	Lists are designed for situations where you need to change data often. So, Python allows editing.
•	Tuples are made for data that should stay constant, which helps keep the program safe and fast.
Example:
my_list = [1, 2, 3]
my_list[0] = 100  # You can change it

my_tuple = (1, 2, 3)
my_tuple[0] = 100  #  Error! You can't change it

8. Logical operators in Python are used to combine conditions (True or False values). They help make decisions in programs.
Here are the three main logical operators, explained simply:
1.	and – Returns True only if both conditions are true.
Example:
                     5 > 2 and 3 < 4  # True because both parts are true
2.	or – Returns True if at least one condition is true.
Example:
                     5 > 2 or 3 > 10  # True because the first part is true
3.	not – Reverses the result.
Example:
                    not (5 > 2)  # False because 5 > 2 is True, and not makes it False

9. Type casting in Python means changing the data type of a value from one type to another.
             For example, turning a number stored as a string into an integer:
                       x = "5"       # x is a string
                       y = int(x)    # y is now an integer
   Common type casting functions:
•	int() – changes to an integer
•	float() – changes to a floating-point number
•	str() – changes to a string
•	bool() – changes to True or False
   Example:
                     a = "10"
                     b = int(a)     # now b is 10 as an integer
                     print(b + 5)   # output: 15

10. Implicit type casting happens automatically by the compiler when it converts one data type     to another (e.g., int to float).
Explicit type casting is done manually by the programmer using type conversion (e.g., float(x) to convert x to float).
Example:
# Implicit
x = 5
y = 2.0
z = x + y   # x is automatically converted to float
# Explicit
a = 5
b = float(a)  # manually converting int to float

11. Conditional statements in Python are used to make decisions in code — they let the program choose different actions based on whether something is true or false.

12. The elif statement means "else if." It checks another condition if the first if condition is false. It helps run different code for different conditions.
Example:
x = 10
if x > 15:
    print("Greater than 15")
elif x > 5:
    print("Greater than 5")
else:
    print("5 or less")
 Since x is 10, the elif is true, so it prints "Greater than 5".

13. A for loop is used when you know how many times you want to repeat something.
A while loop is used when you want to repeat something until a condition is no longer true.
Example:
# for loop - runs 5 times
for i in range(5):
    print(i)
# while loop - runs until i is no longer less than 5
i = 0
while i < 5:
    print(i)
    i += 1

14. A while loop is more suitable than a for loop when you don’t know how many times the loop should run — you just want it to keep running until a certain condition is met.
 Example:
Imagine you're asking a user to enter the correct password.
You don’t know how many times the user will try. So, you use a while loop to keep asking until the correct password is entered.
Example:
password = ""
while password != "1234":
    password = input("Enter your password: ")
print("Access granted!")
# Because with a for loop, you’d have to decide in advance how many attempts to allow. But with a while loop, it continues as long as needed.
